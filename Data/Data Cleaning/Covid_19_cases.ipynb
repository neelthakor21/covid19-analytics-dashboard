### Step 2.2: Make a dictionary for all the countries that have data values spreaded in the different states, so we can identify and combine them under the country name. 

### This code analyzes a dataset (`cases`) where **countries and their states/provinces are stored across columns** rather than in a traditional tabular format.  

### Its main goal is to **identify which countries have state-level data** and to **group all corresponding state columns under each country**.


```python
# Making a dictionary for all the countries that have states mentioned in the dataset.
# This dictionary will be used for removing the state row and combining the data in the country column
def stateFinder(cases):

    # Finding the index of the states whose value is not null.
    notna_indexes = [i for i in cases.columns if pd.notna(cases.iloc[1, i])]
    
    countryWithStates = {}
    for i in notna_indexes:
        Country = cases.iloc[0, i]
        States = []
        for j in cases.columns:
            if Country == cases.iloc[0, j]:
                States.append(j)
        countryWithStates.update({Country: States})

    return countryWithStates

cws_c = stateFinder(cases)
for key, values in cws_c.items():
    print(f"{key} : {values}")
```

### Step 2.3: Now, we can combine all the states' data for each country.

### When aggregating data by country, geographic coordinates (latitude and longitude) are averaged, while all other numeric fields are summed to produce a single consolidated country-level record.

```python
cases.drop('State', axis=0, inplace=True)
```

```python
for country, states in cws_c.items():
    position = cases.iloc[1:3, states].mean(axis=1)
    total = cases.iloc[3:, states].sum(axis=1)
    final = pd.concat([position, total])
    cases.iloc[1:, states[0]] = final

for country, states in cws_c.items():
    cases.drop(states[1:], axis=1, inplace=True)
```

```python
# Make the country name a column name and drop the country row.
cases.columns = cases.iloc[0]
cases.drop('Country', axis=0, inplace=True)
```

```python
def parse_date(date):
    if pd.isna(date):
        return pd.NaT

    date = str(date).strip()

    if '/' in date:
        return pd.to_datetime(date, format="%m/%d/%y", errors="coerce")
    elif '-' in date:
        return pd.to_datetime(date, format="%m-%d-%Y", errors="coerce")
    else:
        return pd.NaT


# cases['index'].iloc[2:] = cases['index'].iloc[2:].apply(parse_date).dt.strftime("%d-%m-%Y")
cases.iloc[2:, 0] = cases.iloc[2:, 0].apply(parse_date).dt.strftime("%d-%m-%Y")
# cases['index'].iloc[2:].apply(parse_date).dt.strftime("%d-%m-%Y")
cases['index']
```
